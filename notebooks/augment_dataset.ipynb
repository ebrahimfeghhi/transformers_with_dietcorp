```python
import pandas as pd

char_dataset = pd.read_pickle('/data/neural_data/ptDecoder_ctc_both_char')

phoneme_dataset = pd.read_pickle('/data/neural_data/ptDecoder_ctc_both')

import copy

# Start with a deep copy of one dataset (to avoid modifying originals)
merged_dataset = copy.deepcopy(char_dataset)

# Iterate over each split: train/test/competition
for split in merged_dataset.keys():
    for i, sample in enumerate(merged_dataset[split]):
        # Add phoneme-related fields from phoneme_dataset
        merged_dataset[split][i]['text2']      = phoneme_dataset[split][i]['text']
        merged_dataset[split][i]['textLens2']     = phoneme_dataset[split][i]['textLens']
        merged_dataset[split][i]['textPerTime2']  = phoneme_dataset[split][i]['textPerTime']

# Now merged_dataset has both text and phoneme info
print(merged_dataset['train'][0].keys())

output_path = "/data/neural_data/ptDecoder_ctc_both_char_phoneme"
pd.to_pickle(merged_dataset, output_path)
```
